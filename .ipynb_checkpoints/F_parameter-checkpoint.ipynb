```python
import numpy as np
import pandas as pd
import os
from scipy.interpolate import interp1d
from tqdm import tqdm

os.environ["PK_MODEL"] = "BCEmu7_high_z"
os.environ["METHOD"] = "exact"
os.environ["N_FRB"]    = "100"
os.environ["N_PARAMS"] = "1"
from mcmc_funcs import *
from astropy.cosmology import Planck18 as cosmo

```

```python
def get_samples(mcmc_file):
    class RenameUnpickler(pickle.Unpickler):
        def find_class(self, module, name):
            if module.startswith("numpy._core"):
                module = module.replace("numpy._core", "numpy.core")
            return super().find_class(module, name)
    with open(mcmc_file, "rb") as f:
        results = RenameUnpickler(f).load()
    chain = results["chain"]
    samples = chain.reshape(-1, chain.shape[-1])
    # DMhost transformations
    DMhost_mean = np.exp(samples[:, -2] + ((samples[:, -1]**2)/2))
    DMhost_var = np.sqrt((np.exp(samples[:, -1]**2) - 1) * np.exp((2*samples[:, -2]) + (samples[:, -1]**2)))
    samples[:, -2] = DMhost_mean; samples[:, -1] = DMhost_var
    return samples


from joblib import Parallel, delayed

def compute_single_sample(i, samples, z_arr, cosmo):
    """Compute DM_var and DM_mean for a single sample."""
    params = np.array([{
        "log_Mc"   : samples[i, 0],
        "theta_ej" : samples[i, 1],
        "eta_delta": 0.2,
        "delta"    : samples[i, 3],
        "mu_beta"  : samples[i, 2],
        "gamma"    : 2.5,
        "eta"      : samples[i, 4],
        "A"        : 0.055 / 2,
        "tau"      : -1.5
    }])

    DM_var = get_DM_variance(
        np.array([cosmo.H0.cgs.value]),
        np.array([cosmo.Ob0]),
        np.array([cosmo.Om0]),
        np.array([0.8111]),
        params,
        z_arr,
        "BCEmu"
    )[0]

    DM_mean = DM_cosmic_mean(
        np.array([cosmo.Om0]),
        np.array([cosmo.Ob0]),
        np.array([cosmo.H0.cgs.value]),
        np.array([0.8111]),
        params,
        "BCEmu",
        z_arr
    )[0]

    return DM_var, DM_mean


BCEmu5_samples = get_samples("data/frb_results/emcee_BCEmu5_extended_Chandraprior.pkl")
z_arr          = np.linspace(0, 3, 100)
N_samp         = 1000

results = Parallel(n_jobs=8, verbose=1)(
    delayed(compute_single_sample)(i, BCEmu5_samples, z_arr, cosmo)
    for i in tqdm(range(N_samp))
)

# unpack
DM_var_arr  = [r[0] for r in results]
DM_mean_arr = [r[1] for r in results]
```

```python
frb_sample = pd.read_csv("data/frb_results/frb_sample.csv")

from scipy.optimize import curve_fit

def compute_F_parameter(z_arr, DM_mean_arr, DM_std_arr, z_frb):
    """
    Compute the Macquart F-parameter from DM mean and variance arrays.
    
    The F-parameter is defined via:
        sigma_DM(z) = F * z^(-1/2)
    where sigma_DM is the relative standard deviation:
        sigma_DM(z) = sqrt(sigma^2(DM_cosmic)) / DM_mean(z)
                    = DM_std(z) / DM_mean(z)
    """
    sigma_DM    = DM_std_arr / DM_mean_arr
    F_of_z = sigma_DM * np.sqrt(z_arr)
    
    frb_counts, bin_edges = np.histogram(z_frb, bins=len(z_arr), range=(z_arr.min(), z_arr.max()))
    bin_centres = 0.5 * (bin_edges[:-1] + bin_edges[1:])
    
    def model(z, F):
        return F * z**(-0.5)
    
    F_best, _ = curve_fit(
        model,
        z_arr,
        sigma_DM,
        p0=[0.3],
        absolute_sigma=False
    )
    F_best = F_best[0]
    
    return F_of_z, F_best

Fs = []
F_of_zs = []
for i in tqdm(range(N_samp)):
    F_of_z, F = compute_F_parameter(z_arr[1:], DM_mean_arr[i][1:], DM_var_arr[i][1:], np.array(frb_sample.z_sample))
    Fs.append(F)
    F_of_zs.append(F_of_z)
```
