```python
import ostrich.emulate
import ostrich.interpolate
import pickle

def make_emulator(parameters, x_qty, y_qty, label, num_components):
    emulator=ostrich.emulate.PcaEmulator.create_from_data(
            parameters,
            y_qty.T,
            ostrich.interpolate.RbfInterpolator,
            interpolator_kwargs={'function':'linear'},
            num_components=num_components)
    with open(label , 'wb') as f:
        pickle.dump((x_qty, emulator), f)

for Ncomp in [5]:
    make_emulator(param_grid, z_obs[1:], np.array(DMvars)[:, 1:], 'emulator_DMvar_{}comp_HMcode.pkl'.format(Ncomp), Ncomp)
```

```python
with open('emulator_DMvar_5comp_HMcode.pkl', 'rb') as f:
    _, emulator_DMvars = pickle.load(f)

fig = plt.figure(figsize=(8, 7)); grid = plt.GridSpec(4, 4, hspace=0.45, wspace=0.6); ax = fig.add_subplot(grid[:-1, 1:])
    
err_samples = []
for i in range(3125):
    truth = np.reshape(DMvars[i], (100, 1))[1:, :]
    emulation = emulator_DMvars([param_grid[i]])
    err_samples.append(np.abs((truth-emulation)/truth))
    
ax.fill_between(z_obs[1:], 0, np.percentile(err_samples, 68, axis=0)[:, 0], color="royalblue", alpha=0.75)
ax.fill_between(z_obs[1:], 0, np.percentile(err_samples, 95, axis=0)[:, 0], color="royalblue", alpha=0.5)
ax.fill_between(z_obs[1:], 0, np.percentile(err_samples, 99, axis=0)[:, 0], color="royalblue", alpha=0.25)

ax.set_xlabel(r"$z$"); ax.set_ylabel(r"$\frac{\sigma(\mathrm{DM}|z)^\mathrm{true}-\sigma(\mathrm{DM}|z)^\mathrm{emulated}}{\sigma(\mathrm{DM}|z)^\mathrm{true}}$")
ax.set_xlim(z_obs[0], 1.5)
```

```python
import os; os.chdir("../"); from mcmc_funcs import *; os.chdir("HMcode")

z = np.linspace(0, 2, 100)
DMvar = get_DM_variance(cosmo.H0.cgs.value, cosmo.Ob0, cosmo.Om0, 0.8111, {"Tagn": 7.8}, z, Pk_model="HMcode", method="emulate")
DMmean = DM_cosmic_mean(cosmo.Om0, cosmo.Ob0, cosmo.H0.cgs.value, z)

Nsamples = 10000
z_sampled = np.random.uniform(0, 1.4, Nsamples)

def sample_lognormal(redshift):
    mean = np.interp(redshift, z, DMmean)
    variance = np.interp(redshift, z, DMvar)**2
    
    mu_log = np.log(mean**2 / np.sqrt(mean**2 + variance))
    sigma_log = np.sqrt(np.log(1 + variance / mean**2))
    sample = lognorm.rvs(s=sigma_log, scale=np.exp(mu_log), size=1)[0]
    return sample

DM_sampled = []
for i in tqdm(range(Nsamples)):
    DM_sampled.append(sample_lognormal(z_sampled[i]))
DM_sampled = np.array(DM_sampled)

host_sigma, host_mu = 0.5, 5
DMhost = lognorm.rvs(s=host_sigma, scale=np.exp(host_mu), size=Nsamples)
DMexgal = DM_sampled + DMhost/(1+z_sampled) + 50

df = pd.DataFrame({"DMexgal": DMexgal, "DMhost": DMhost, "DMigm": DM_sampled, "z_sample": z_sampled})

df.to_csv("HMcode_DMsamples.csv")
df.head()
```

```python
with open('emulator_DMvar_5comp_HMcode.pkl', 'rb') as f:
    _, emulator_DMvars = pickle.load(f)

variation1 = cosmo.H0.value-1
variation2 = cosmo.H0.value+1
DMvar_varied1 = emulator_DMvars([[variation1, cosmo.Ob0, cosmo.Om0, 0.8111, 7.8]])
DMvar_varied2 = emulator_DMvars([[variation2, cosmo.Ob0, cosmo.Om0, 0.8111, 7.8]])
dDMvar_dH0 = ((DMvar_varied2 - DMvar_varied1) / (variation2-variation1))/(((DMvar_varied2 + DMvar_varied1))/2)

variation1 = cosmo.Ob0-0.01
variation2 = cosmo.Ob0+0.01
DMvar_varied1 = emulator_DMvars([[cosmo.H0.value, variation1, cosmo.Om0, 0.8111, 7.8]])
DMvar_varied2 = emulator_DMvars([[cosmo.H0.value, variation2, cosmo.Om0, 0.8111, 7.8]])
dDMvar_Ob0 = ((DMvar_varied2 - DMvar_varied1) / (variation2-variation1))/(((DMvar_varied2 + DMvar_varied1))/2)

variation1 = cosmo.Om0-0.1
variation2 = cosmo.Om0+0.1
DMvar_varied1 = emulator_DMvars([[cosmo.H0.value, cosmo.Ob0, variation1, 0.8111, 7.8]])
DMvar_varied2 = emulator_DMvars([[cosmo.H0.value, cosmo.Ob0, variation2, 0.8111, 7.8]])
dDMvar_Om0 = ((DMvar_varied2 - DMvar_varied1) / (variation2-variation1))/(((DMvar_varied2 + DMvar_varied1))/2)

variation1 = 0.8111-0.05
variation2 = 0.8111+0.05
DMvar_varied1 = emulator_DMvars([[cosmo.H0.value, cosmo.Ob0, cosmo.Om0, variation1, 7.8]])
DMvar_varied2 = emulator_DMvars([[cosmo.H0.value, cosmo.Ob0, cosmo.Om0, variation2, 7.8]])
dDMvar_s8 = ((DMvar_varied2 - DMvar_varied1) / (variation2-variation1))/(((DMvar_varied2 + DMvar_varied1))/2)

variation1 = 7.8-0.1
variation2 = 7.8+0.1
DMvar_varied1 = emulator_DMvars([[cosmo.H0.value, cosmo.Ob0, cosmo.Om0, 0.8111, variation1]])
DMvar_varied2 = emulator_DMvars([[cosmo.H0.value, cosmo.Ob0, cosmo.Om0, 0.8111, variation2]])
dDMvar_Tagn = ((DMvar_varied2 - DMvar_varied1) / (variation2-variation1))/(((DMvar_varied2 + DMvar_varied1))/2)
```

```python
DMvar_arr = [dDMvar_dH0[:, 0], dDMvar_Ob0[:, 0], dDMvar_Om0[:, 0], dDMvar_s8[:, 0], dDMvar_Tagn[:, 0]]
cov_matrix = np.zeros((5, 5))
for i in range(len(DMvar_arr)):
    for j in range(len(DMvar_arr)):
        cov_matrix[i, j] = np.dot(DMvar_arr[i], DMvar_arr[j])/(np.linalg.norm(DMvar_arr[i])*np.linalg.norm(DMvar_arr[j]))

param_labels = [r'$H_0$', r"$\Omega_\mathrm{b}$", r"$\Omega_\mathrm{m}$", r"$\sigma_8$", r'$T_{\mathrm{AGN}}$']

import seaborn as sns
plt.figure(figsize=(6,5))
sns.heatmap(cov_matrix, annot=True, cmap="seismic", fmt=".2f", xticklabels=param_labels, yticklabels=param_labels, 
            cbar_kws={'label': 'Correlation Coefficient'})
```
