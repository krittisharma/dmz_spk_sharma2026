```python
mcmc_file = 'emcee_1000FRBs_HMcode.pkl'

ndim = len(mins)
if __name__ == '__main__':
    moves = [(emcee.moves.StretchMove(a=1.5), 0.6),
             (emcee.moves.KDEMove(), 0.4)]

    nwalkers = 4*ndim
    burn = 1000
    steps = 10000

    map_estimate = np.array([cosmo.Ob0, cosmo.Om0, cosmo.H0.value, 0.8111, 7.8, 5, 0.5])
    p0_x = map_estimate + 1e-3 * np.random.randn(nwalkers, ndim)

    with Pool(processes=8) as pool:
        sampler = emcee.EnsembleSampler(nwalkers, ndim, log_probability, moves=moves, pool=pool)

        state = sampler.run_mcmc(p0_x, burn, progress=True)
        sampler.reset()
        sampler.run_mcmc(state, steps, progress=True)

    results = {
        "chain": sampler.get_chain(),
        "log_prob": sampler.get_log_prob(),
        "acceptance_fraction": sampler.acceptance_fraction,
    }

    with open(mcmc_file, "wb") as f:
        pickle.dump(results, f)

```

```python
k_fields = np.geomspace(1e-4, 1e2, 1000)
z = np.linspace(0, 1, 100)

def get_sampled_SPks_DMvars(mcmc_file=None, resampled_samples=None):
    if mcmc_file != None:
        with open(mcmc_file, "rb") as f:
            results = pickle.load(f)
        chain = results["chain"]; nsteps, nwalkers, ndim = chain.shape; resampled_samples = chain.reshape(-1, ndim)

    from joblib import Parallel, delayed
    def compute_SPk(s):
        return get_SPk_HMcode(s[2]*1e5/Mpc, s[0], s[1], s[3], s[4], method="emulate")
    sampled_SPks = Parallel(n_jobs=-1, backend="loky")(
        delayed(compute_SPk)(resampled_samples[i]) for i in tqdm(range(240000))
    )
    def compute_DMvar(s):
        return get_DM_variance(s[2]*1e5/Mpc, s[0], s[1], s[3], {"Tagn": s[4]}, z, method="emulate", Pk_model="HMcode")
    sampled_DMvars = Parallel(n_jobs=-1, backend="loky")(
        delayed(compute_DMvar)(resampled_samples[i]) for i in tqdm(range(240000))
    )
    return np.array(sampled_SPks), np.array(sampled_DMvars)


sampled_SPks_100, sampled_DMvars_100 = get_sampled_SPks_DMvars('emcee_100FRBs_HMcode.pkl')
sampled_SPks_1000, sampled_DMvars_1000 = get_sampled_SPks_DMvars('emcee_1000FRBs_HMcode.pkl')

resampled_samples = np.array([prior(np.random.uniform(size=7)) for i in range(240000)])
sampled_SPks_prior, sampled_DMvars_prior = get_sampled_SPks_DMvars(resampled_samples=resampled_samples)
```

```python
def plot_with_posteriors(ax, x, samples, color, f=16, label=None):
    """Plot mean curve and posterior intervals for a set of samples."""
    mean_curve = np.mean(samples, axis=0)
    ax.plot(x, mean_curve, color=color, ls="solid", lw=1.5)
    
    lower = np.percentile(samples, f, axis=0)
    upper = np.percentile(samples, 100-f, axis=0)
    ax.fill_between(x, lower, upper, color=color, alpha=0.4)


fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].plot(
    k_fields, 
    get_SPk_HMcode(cosmo.H0.cgs.value, cosmo.Ob0, cosmo.Om0, 0.8111, 7.8, method="emulate"), 
    color='black', ls="--", lw=1.5
)

plot_with_posteriors(axes[0], k_fields, sampled_SPks_prior, color="gray", f = 0)
plot_with_posteriors(axes[0], k_fields, sampled_SPks_100, color="red")
plot_with_posteriors(axes[0], k_fields, sampled_SPks_1000, color="blue")

axes[0].set_xscale('log')
axes[0].axhline(1, ls="solid", color="black")
axes[0].set_xlim(1e-2, 1e2); axes[0].set_ylim(0, 2)
axes[0].set_xlabel(r"$k~[h~\mathrm{Mpc}^{-1}]$"); axes[0].set_ylabel(r"$P_\mathrm{hydro}/P_\mathrm{gravity-only}$")


axes[1].plot(
    z, 
    get_DM_variance(cosmo.H0.cgs.value, cosmo.Ob0, cosmo.Om0, 0.8111, {"Tagn": 7.8}, z, method="emulate", Pk_model="HMcode"), 
    color='black', ls="--", lw=1.5
)

plot_with_posteriors(axes[1], z, sampled_DMvars_prior, color="gray", f = 0)
plot_with_posteriors(axes[1], z, sampled_DMvars_100, color="red")
plot_with_posteriors(axes[1], z, sampled_DMvars_1000, color="blue")

axes[1].set_xlabel(r"$z$"); axes[1].set_ylabel(r"$\sigma(\mathrm{DM}|z)~[\mathrm{pc}~\mathrm{cm}^{-3}]$")
axes[1].set_xlim(0, 1); axes[1].set_ylim(0, 250)

plt.tight_layout()
plt.show()
```
